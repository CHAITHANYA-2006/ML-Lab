```python
states = ['Walking', 'Running', 'Resting']
observations = ['High', 'Medium', 'Low', 'High']

pi = {
    'Walking': 0.5,
    'Running': 0.3,
    'Resting': 0.2
}

A = {
    'Walking': {'Walking': 0.6, 'Running': 0.3, 'Resting': 0.1},
    'Running': {'Walking': 0.2, 'Running': 0.5, 'Resting': 0.3},
    'Resting': {'Walking': 0.3, 'Running': 0.2, 'Resting': 0.5}
}

B = {
    'Walking': {'Low': 0.6, 'Medium': 0.3, 'High': 0.1},
    'Running': {'Low': 0.1, 'Medium': 0.3, 'High': 0.6},
    'Resting': {'Low': 0.7, 'Medium': 0.2, 'High': 0.1}
}
```

```python
alpha = {}

# Step 1 : Initialization
alpha[0] = {}
for state in states:
    alpha[0][state] = pi[state] * B[state][observations[0]]
print("After Initialization:")
print(alpha[0])

# Step 2 : Recursion
for t in range(1, len(observations)):
    alpha[t] = {}
    for curr_state in states:
        total = 0
        for prev_state in states:
            total += alpha[t-1][prev_state] * A[prev_state][curr_state]
        alpha[t][curr_state] = total * B[curr_state][observations[t]]
    print(f"\nAfter Observation {t+1}:")
    print(alpha[t])

# Step 3 : Termination
probability = sum(alpha[len(observations)-1].values())
print("\nForward Probability Table:")
for t in alpha:
    print(f"Time {t+1} :", alpha[t])
print("\nLikelihood P(O|lam) =", probability)
```

```python
import numpy as np
from hmmlearn.hmm import CategoricalHMM

model = CategoricalHMM(n_components=3)
model.startprob_ = np.array([0.5, 0.3, 0.2])
model.transmat_ = np.array([
    [0.6, 0.3, 0.1],
    [0.2, 0.5, 0.3],
    [0.3, 0.2, 0.5]
])
model.emissionprob_ = np.array([
    [0.6, 0.3, 0.1],   # Walking
    [0.1, 0.3, 0.6],   # Running
    [0.7, 0.2, 0.1]    # Resting
])

# Low = 0 Medium = 1 High = 2
O  = np.array([[2], [1], [0], [2]])
O1 = np.array([[0], [1], [2], [1]])
O2 = np.array([[2], [2], [2], [2]])
O3 = np.array([[0], [0], [0], [0]])
O4 = np.array([[1], [1], [1], [1]])


print("Likelihood of O  =", np.exp(model.score(O)))
print("Likelihood of O1 =", np.exp(model.score(O1)))
print("Likelihood of O2 =", np.exp(model.score(O2)))
print("Likelihood of O3 =", np.exp(model.score(O3)))
print("Likelihood of O4 =", np.exp(model.score(O4)))
```
